# Parameter Space Explorer

Generate the full combinatorial grid of simulation parameters and export it as a
`DataFrame` for inspection.  Use this to plan which parameter combinations to run
with `main.py`, or to build a batch-run script.

Each row in the resulting table represents one unique simulation configuration.
The `sim_id` column (`sim_00001`, `sim_00002`, …) can be used as the experiment
name in `main.py`.


```python
import itertools
import pandas as pd

# ── Sweep parameters (each list is one axis of the grid) ──────────────────
params = {
    "tree_shape":              ["regular", "triangular"],
    "branch_arrangement":      ["staggered", "aligned"],
    "number_of_layers":        [2, 6, 12],
    "chitin_layer_thickness":  [0.05, 0.1, 0.3],
    "air_layer_thickness":     [0.01, 0.1, 0.4],
    "air_gap_x":               [0.05, 0.1, 0.15, 0.2, 0.25, 0.3,
                                 0.35, 0.4, 0.45, 0.5, 1.0, 1.5, 2.0],
    "ridge_total_width":       [0.24, 0.4, 0.6, 1.0, 1.5],
}

# ── Fixed parameters (constant across all combinations) ───────────────────
fixed_params = {
    "substrate_thickness_z": 0.164,
    "air_gap_below_z":       1.8,
    "ridge_trunk_width":     0.12,
}

# ── Build full combinatorial grid ─────────────────────────────────────────
keys, values = zip(*params.items())
combinations = [dict(zip(keys, v)) for v in itertools.product(*values)]

df = pd.DataFrame(combinations)

# Append fixed columns and a zero-padded simulation ID
for k, v in fixed_params.items():
    df[k] = v

df["sim_id"] = df.index.map(lambda x: f"sim_{x:05d}")

print(f"Total combinations: {len(df):,}")
```
